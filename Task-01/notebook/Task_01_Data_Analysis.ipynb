# TIME-BASED ANALYSIS

How many games were released each year?

```python
games_per_year = df_year.groupby("year")["name"].count().reset_index(name="game_count")
```

```python
games_per_year.head()
```

#### Plot:

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=games_per_year,
    x="year",
    y="game_count"
)

plt.title("Number of Video Games Released Over Time")
plt.xlabel("Release Year")
plt.ylabel("Number of Games")

plt.show()
```

**Interpretation:** The number of releases changes substantially across the years and reaches its highest level around 2008–2009. This shows a period of particularly high release activity in the dataset.


>This directly gives us a time-based trend.

***********

How did global sales change over time?

```python
sales_per_year = (
    df_year.groupby("year")["global_sales"]
    .sum()
    .reset_index()
)
```

```python
sales_per_year.head()
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=sales_per_year,
    x="year",
    y="global_sales"
)

plt.title("Global Video Game Sales Over Time")
plt.xlabel("Release Year")
plt.ylabel("Global Sales (Millions)")

plt.show()
```

**Interpretation:** Total global sales vary considerably by year. Years with more releases can naturally have higher total sales, so this chart is most useful for identifying broad changes in total market activity rather than the performance of an individual game.


What was the average global sales per game over time?

```python
avg_sales_per_year = (
    df_year.groupby("year")["global_sales"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=avg_sales_per_year,
    x="year",
    y="global_sales"
)

plt.title("Average Global Sales per Game Over Time")
plt.xlabel("Release Year")
plt.ylabel("Average Global Sales (Millions)")

plt.show()
```

**Interpretation:** Average sales per game help separate the effect of release volume from the effect of individual game performance. Large changes in the average indicate years where games tended to perform differently on a per-title basis.


>year with many games can naturally have high total sales.

>We'll use genre distribution.

#  HISTOGRAM:

>understand global sales distribution.

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="global_sales",
    bins=50
)

plt.title("Distribution of Global Video Game Sales")
plt.xlabel("Global Sales (Millions)")
plt.ylabel("Number of Games")

plt.show()
```

# SCATTER PLOT:

>We can investigate whether North American sales are related to European sales.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="na_sales",
    y="eu_sales"
)

plt.title("Relationship Between North American and European Sales")
plt.xlabel("North American Sales (Millions)")
plt.ylabel("European Sales (Millions)")

plt.show()
```

**Interpretation:** The scatter plot shows a clear positive relationship between North American and European sales. Games that sell well in North America often also sell well in Europe.


# CORRELATION HEATMAP:

```python
correlation = df[sales_columns].corr()
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Between Regional and Global Sales")

plt.show()
```

**Interpretation:** North American and European sales have especially strong correlations with global sales, while Japanese sales have a weaker but still positive relationship.


## COUNT PLOT:

>Top 10 platforms by number of games

```python
top_platforms = df["platform"].value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df[df["platform"].isin(top_platforms.index)],
    y="platform",
    order=top_platforms.index
)

plt.title("Top 10 Platforms by Number of Games")
plt.xlabel("Number of Games")
plt.ylabel("Platform")

plt.show()
```

## VIOLIN PLOT:

```python
top_genres = df["genre"].value_counts().head(6).index
genre_data = df[df["genre"].isin(top_genres)]
plt.figure(figsize=(12, 6))

sns.violinplot(
    data=genre_data,
    x="genre",
    y="global_sales"
)

plt.title("Global Sales Distribution Across Major Genres")
plt.xlabel("Genre")
plt.ylabel("Global Sales (Millions)")

plt.show()
```

**Interpretation:** The violin plot compares the distribution of global sales across the most common genres. The width and shape show where observations are concentrated and how spread differs between genres.


## PAIR PLOT:

```python
pair_data = df[
    ["na_sales", "eu_sales", "jp_sales", "other_sales", "global_sales"]
]

sns.pairplot(pair_data.sample(1000,random_state=42))

plt.show()
```

**Interpretation:** The pair plot compares regional and global sales together. The positive patterns across several panels are consistent with the correlation results and show that stronger regional performance is generally associated with stronger global performance.
